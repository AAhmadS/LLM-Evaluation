```python
class Metric:
    def __init__(self, name, description, scale, instruction=""):
        self.name = name
        self.description = description
        self.scale = scale
        self.instruction = instruction

    def set_instruction(self, instruction):
        self.instruction = instruction
    def to_str(self):
        return f"[Metric]: Name: {self.name}, Description: {self.description}, Scale: {self.scale} [Metric]\n"
    def to_inst(self):
        return  f"[Metric]: Name: {self.name}, Description: {self.description}, Scale: {self.scale} [Metric]\n{self.instruction}"

def serialize_metric(metric):
    return {
        "name": metric.name,
        "description": metric.description,
        "scale": metric.scale,
        "instruction": metric.instruction
    }

def deserialize_metric(metric_dict):
    return Metric(**metric_dict)

class Metrics_set:
    def __init__(self, dataset, source, set_of_metrics):
        self.dataset = dataset
        self.source = source
        self.set_of_metrics = set_of_metrics
        self.overall_metric = set_of_metrics[-1]
```

```python
from docx import Document
import json
# Function to filter metrics based on unique names
def filter_metrics(metrics):
    unique_metrics = {}
    for metric in metrics:
        if metric.name not in unique_metrics:
            unique_metrics[metric.name] = metric
    return list(unique_metrics.values())

for model_name in ["4o-mini", "4o", "llama", "mixtral", "qwen"]:
    doc = Document()
    files_list = model_files[model_name]
    # Open a text file for writing
    with open(f'{model_name}-filtered_metrics.txt', 'w') as txt_file:
        for filename in files_list:
            # Load metrics from the file
            metrics = json.load(open(filename, 'r'))
            metrics = [deserialize_metric(metric) for pack in metrics for metric in pack]

            # Filter metrics to remove redundancies
            filtered_metrics = filter_metrics(metrics)

            # Write to the text file
            txt_file.write(f"Filename: {filename}\n")
            for metric in filtered_metrics:
                txt_file.write(metric.to_str() + "\n")
            txt_file.write("\n")

            # Write to the .docx file
            doc.add_heading(f"Filename: {filename}", level=1)
            for metric in filtered_metrics:
                doc.add_paragraph(metric.to_str())
            doc.add_paragraph("\n")

    # Save the .docx file
    doc.save(f'{model_name}-filtered_metrics.docx')
```

```python
import numpy as np
import seaborn as sns
from sklearn.metrics.pairwise import cosine_similarity
import json
import pickle

import matplotlib.pyplot as plt
import torch

def plot_similarity_matrix(setting_name, metrics_list2, description2, embeddings2, names2, model, dataset_name):
    # Extract descriptions
    metrics_list1 = [deserialize_metric(metric) for metric in json.load(open(f'definition/{dataset_name}/{setting_name}'))[0]]
    descriptions1 = [metric.description for metric in metrics_list1]
    names1 = [metric.name for metric in metrics_list1]

    # Get embeddings
    with torch.no_grad():
      embeddings1 = model.encode(descriptions1, task="text-matching", truncate_dim=256)

    # Calculate cosine similarity
    similarity_matrix = cosine_similarity(embeddings1, embeddings2)

    # Plot the similarity matrix
    plt.figure(figsize=(10, 8))
    sns.heatmap(similarity_matrix, annot=True, cmap='coolwarm', xticklabels=names2, yticklabels=names1)
    plt.title('Cosine Similarity Matrix')
    plt.xlabel('Metrics List 2')
    plt.ylabel('Metrics List 1')
    plt.xticks(rotation=45)
    plt.yticks(rotation=45)

    # Save the plot
    plt.savefig(f'definition/{dataset_name}/Similarities/{setting_name}.png')
    plt.close()

# Example usage
# plot_similarity_matrix(filtered_metrics, filtered_metrics, model, 'output_directory')
```

```python
import os

def remove_all_files_in_folder(folder_path):
    """
    Remove all files in the specified folder.
    :param folder_path: Path to the folder.
    """
    if not os.path.exists(folder_path):
        print(f"The folder {folder_path} does not exist.")
        return

    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)
        try:
            if os.path.isfile(file_path):
                os.remove(file_path)
                print(f"Deleted file: {file_path}")
        except Exception as e:
            print(f"Error deleting file {file_path}: {e}")

# Example usage
folder_path = "definition/USR/Similarities"  # Replace with your folder path
remove_all_files_in_folder(folder_path)
```

```python
import shutil

def zip_folder(folder_path, output_path):
    """
    Compresses a folder into a zip file.

    :param folder_path: Path to the folder to be compressed
    :param output_path: Path for the output zip file (without the .zip extension)
    """
    try:
        shutil.make_archive(output_path, 'zip', folder_path)
        print(f"Folder successfully zipped to {output_path}.zip")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage:
# Replace 'your_folder_path' with the path to the folder you want to zip.
# Replace 'your_output_path' with the desired path for the output zip file.
folder_to_zip = "definition/USR/Similarities"
output_zip = "Similarities"

zip_folder(folder_to_zip, output_zip)
```

```python
import numpy as np
import seaborn as sns
from sklearn.metrics.pairwise import cosine_similarity
import json
import pickle

import matplotlib.pyplot as plt
import torch

def calculate_similarity(setting_name, metrics_list2, description2, embeddings2, names2, model, dataset_name):
    # Extract descriptions
    metrics_list1 = [deserialize_metric(metric) for metric in json.load(open(f'definition/{dataset_name}/{setting_name}'))[0]]
    descriptions1 = [metric.description for metric in metrics_list1]
    names1 = [metric.name for metric in metrics_list1]

    # Get embeddings
    with torch.no_grad():
      embeddings1 = model.encode(descriptions1, task="text-matching", truncate_dim=256)

    # Calculate cosine similarity
    similarity_matrix = cosine_similarity(embeddings1, embeddings2)

    # Threshold for matching
    threshold = 0.5

    # Initialize counters and variables
    covered = 0
    similarities_sum = 0.0
    matched_indices = []

    # Check matches for each item in descriptions2
    for i, row in enumerate(similarity_matrix.T):  # Transpose to compare descriptions2 against descriptions1
        max_similarity = np.max(row)
        if max_similarity >= threshold:
            covered += 1
            similarities_sum += (max_similarity + row[np.argmax(row)]) / 2  # Mean similarity with the most similar item
            matched_indices.append(i)

    # Calculate mean of similarities for matched items
    mean_of_similarities = similarities_sum / covered if covered > 0 else 0.0

    # Reverse coverage: Check matches for each item in descriptions1
    reverse_covered = 0
    for row in similarity_matrix:
        if np.max(row) >= threshold:
            reverse_covered += 1

    # Determine if all items in descriptions2 have a match with similarity >= threshold
    satisfied = (covered == len(descriptions2))

    # Output the results
    return (satisfied, covered, reverse_covered, mean_of_similarities)

# Example usage
# plot_similarity_matrix(filtered_metrics, filtered_metrics, model, 'output_directory')
```
